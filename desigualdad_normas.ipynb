#### $\alpha=2$ and $\beta=1$ case

To plot the unit circle in 1-norm in the positive quadrant, we must solve:
$$
\|x\|_1 = |x| + |y| = 1
$$
As $x, y \geq 0$, we can drop the absolute values and solve for $y$:
$$
y = 1 - x
$$
Now, as $x$ varies from $0$ to $1$, we can plot the unit circle in 1-norm in the positive quadrant.

For the unit circle in 2-norm, we have:
$$
\|x\|_2 = \sqrt{x^2 + y^2} = 1
$$
Again, as $x, y \geq 0$, we can drop the absolute values and solve for $y$:
$$
y = \sqrt{1 - x^2}
$$

```python
X = np.linspace(0,1, 100)

Y_1 = 1 - X
Y_2 = np.sqrt(1 - X**2)

plt.figure(figsize=(8, 8))
plt.plot(X, Y_1, label='1-norm', color='blue')
plt.plot(X, Y_2, label='2-norm', color='orange')

plt.gca().set_aspect('equal', adjustable='box')
plt.axhline(0, color='black',linewidth=1.0, ls='-')
plt.axvline(0, color='black',linewidth=1.0, ls='-')
plt.grid(color = 'gray', linestyle = '--', linewidth = 0.5)

plt.title('1-norm and 2-norm circles in 2D')
plt.xlabel('x')
plt.ylabel('y')
plt.legend()

plt.show()
```

Now if we take the points on the 2-norm unit circle, we can find the maximum and minimum values of the 1-norm for these points. With some algebraic manipulation, we can find that the maximum value is $\frac{1}{\sqrt{2}}$ and the minimum value is $0$ or $1$.

```python
NORM_1 = lambda x,y: np.abs(x) + np.abs(y)
NORM_2 = lambda x,y: np.sqrt(x**2 + y**2)

CIRCLE_2_NORM_1 = NORM_1(X, Y_2)
CIRCLE_2_NORM_2 = NORM_2(X, Y_2)

CIRCLE_2_NORM_1_MAX_X = np.sqrt(2)/2
CIRCLE_2_NORM_1_MAX_Y = NORM_1(CIRCLE_2_NORM_1_MAX_X, CIRCLE_2_NORM_1_MAX_X)

plt.figure(figsize=(8, 8))

plt.axhline(0, color='black',linewidth=1.0, ls='-')
plt.axvline(0, color='black',linewidth=1.0, ls='-')
plt.grid(color = 'gray', linestyle = '--', linewidth = 0.5)

plt.plot(X, CIRCLE_2_NORM_1, label='1-norm for points in 2-norm unit circle', color='blue')
plt.plot(X, CIRCLE_2_NORM_2, label='2-norm for points in 2-norm unit circle', color='orange')

plt.plot([0,1], [1,1], 'go', label='Min 1-norm point')
plt.plot(CIRCLE_2_NORM_1_MAX_X, CIRCLE_2_NORM_1_MAX_Y, 'ro', label='Max 1-norm point')

plt.gca().set_aspect('equal', adjustable='box')

plt.title('1-norm and 2-norm circles in 2D')
plt.xlabel('x')
plt.ylabel('y')
plt.legend()

plt.show()
```

Now, for $c_1 \cdot \|x\|_1 \leq \|x\|_2 \leq c_2 \cdot \|x\|_1$, as we are working with $x: \|x\|_2 = 1$, we want $c_1$ such that:
$$
\max_x c_1 \cdot \|x\|_1 = 1, \quad \|x\|_2 = 1
$$
and a $c_2$ such that:
$$
\min_x c_2 \cdot \|x\|_1 = 1, \quad \|x\|_2 = 1
$$
With $\alpha = 2, \beta = 1$ we got, for $c_1$:
$$
\begin{align*}
\max_x c_1 \cdot \|x\|_1 &= 1, \quad \|x\|_2 = 1 \\
c_1 \cdot \left\|\begin{pmatrix} \frac{\sqrt{2}}{2} \\ \frac{\sqrt{2}}{2} \end{pmatrix}\right\|_1 &= 1 \\
c_1 \cdot \frac{\sqrt{2}}{2} + \frac{\sqrt{2}}{2}&= 1 \\
c_1 \cdot \sqrt{2}&= 1 \\
c_1 &= \frac{\sqrt{2}}{2} \\
\end{align*}
$$
and for $c_2$,
$$
\begin{align*}
\min_x c_2 \cdot \|x\|_1 &= 1, \quad \|x\|_2 = 1 \\
c_2 \cdot \left\|\begin{pmatrix} 0 \\ 1 \end{pmatrix}\right\|_1 &= 1 \\
c_2 \cdot 0 + 1 &= 1 \\
c_2 &= 1 \\
\end{align*}
$$

```python
X = np.linspace(0,1, 100)

Y_1 = 1 - X
Y_2 = np.sqrt(1 - X**2)

C1 = 1
C2 = np.sqrt(2)

plt.figure(figsize=(8, 8))
plt.plot(X, Y_1, label='1-norm', color='blue')
plt.plot(X, Y_2, label='2-norm', color='orange')
plt.plot(C1 * X, C1 * Y_1, label='1-norm (c1 scaled)', color='green', ls='--')
plt.plot(C2 * X, C2 * Y_1, label='1-norm (c2 scaled)', color='red', ls='--')

plt.gca().set_aspect('equal', adjustable='box')
plt.axhline(0, color='black',linewidth=1.0, ls='-')
plt.axvline(0, color='black',linewidth=1.0, ls='-')
plt.grid(color = 'gray', linestyle = '--', linewidth = 0.5)

plt.title('1-norm and 2-norm circles in 2D')
plt.xlabel('x')
plt.ylabel('y')
plt.legend()

plt.show()
```

##### But, ¿How do we generalize for $n \neq 2$? 
> Por mucho que me gustó el ejemplo geométrico, para demostrarlo en $n \neq 2$, debemos abandonarlo...

```python
from ipywidgets import interact, FloatSlider, ToggleButton
import sympy as sp

def norms(alpha, beta):
    n = 2
    c1, c2 = (1/n**(1/alpha - 1/beta), 1) if alpha < beta else (1, 1/n**(1/alpha - 1/beta))

    parametrization = lambda t, p: (1 - np.abs(t)**p)**(1/p)

    num_points = 100
    x = np.zeros((4*num_points, 3))
    x[:, 0] = np.concatenate([
        np.linspace(0, 1, num_points), 
        np.linspace(1, 0, num_points), 
        np.linspace(0, -1, num_points), 
        np.linspace(-1, 0, num_points)
    ])
    x[:, 1] = np.concatenate([
        parametrization(x[:num_points, 0], alpha),
        -parametrization(x[num_points:2*num_points, 0], alpha),
        -parametrization(x[2*num_points:3*num_points, 0], alpha),
        parametrization(x[3*num_points:, 0], alpha)
    ])

    x[:, 2] = np.concatenate([
        parametrization(x[:num_points, 0], beta),
        -parametrization(x[num_points:2*num_points, 0], beta),
        -parametrization(x[2*num_points:3*num_points, 0], beta),
        parametrization(x[3*num_points:, 0], beta)
    ])

    print(f"c1: {c1}, c2: {c2}")

    plt.figure(figsize=(8, 8))

    plt.gca().set_aspect('equal', adjustable='box')
    plt.axhline(0, color='black',linewidth=1.0, ls='-')
    plt.axvline(0, color='black',linewidth=1.0, ls='-')
    plt.grid(color = 'gray', linestyle = '--', linewidth = 0.5)

    plt.plot(x[:, 0], x[:, 1], label=f'Unit {alpha}-norm', color='blue')
    plt.plot(c1*x[:, 0], c1*x[:, 2], label=f'Unit {beta}-norm (c1 scaled)', color='green')
    plt.plot(c2*x[:, 0], c2*x[:, 2], label=f'Unit {beta}-norm (c2 scaled)', color='red')

    plt.legend()

    plt.show()

@interact(
    alpha=FloatSlider(value=1, min=0.1, max=10, step=0.1), 
    beta=FloatSlider(value=2, min=0.1, max=10, step=0.1)
)
def update(alpha, beta):
    if alpha == 10.0:
        alpha = np.inf
    if beta == 10.0:
        beta = np.inf
    norms(alpha, beta)
```
